# Fill missing marks with 0

```python
import pandas as pd

df = pd.DataFrame({
    "Name": [
        "John",
        "Anu",
        "Rahul",
        "Meera",
        "Arjun",
        "Diya",
        "Vishnu",
        "Neha"
    ],
    "Age": [
        21,
        22,
        None,
        20,
        23,
        None,
        21,
        22
    ],
    "Course": [
        "Python",
        "SQL",
        "Python",
        None,
        "ML",
        "SQL",
        None,
        "Python"
    ],
    "Marks": [
        85,
        None,
        72,
        90,
        None,
        68,
        95,
        None
    ]
})
```

```python
df
```

fillna(0) is used to replace missing values (NaN) with 0 in a Pandas DataFrame or Series.

```python
df['Marks'] =df['Marks'].fillna(0)
```

# Filling Missing marks with Mean



```python
import pandas as pd

df = pd.DataFrame({
    "Name": [
        "John",
        "Anu",
        "Rahul",
        "Meera",
        "Arjun",
        "Diya",
        "Vishnu",
        "Neha"
    ],
    "Age": [
        21,
        22,
        None,
        20,
        23,
        None,
        21,
        22
    ],
    "Course": [
        "Python",
        "SQL",
        "Python",
        None,
        "ML",
        "SQL",
        None,
        "Python"
    ],
    "Marks": [
        85,
        None,
        72,
        90,
        None,
        68,
        95,
        None
    ]
})
```

```python
df
```

```python
df['Marks']=df['Marks'].fillna(df['Marks'].mean())
df['Marks']
```
